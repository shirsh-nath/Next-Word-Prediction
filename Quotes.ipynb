```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import string

import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,LSTM,Dense
```

```python
df=pd.read_csv("/content/qoute_dataset.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
quotes=df["quote"]
quotes
```

```python
quotes.apply(lambda x:x.translate(str.maketrans("","",string.punctuation)))
quotes.head()
```

```python
quotes=quotes.str.lower()
```

```python
vocab_size=10000
tokenizer=Tokenizer(num_words=vocab_size,oov_token="<oov>")
tokenizer.fit_on_texts(quotes)

word_index=tokenizer.word_index
print(len(word_index))
list(word_index.items())[:9]
```

```python
sequences=tokenizer.texts_to_sequences(quotes)
sequences[0]
```

```python
X=[]
Y=[]
for sequence in sequences:
  for i in range(1,len(sequence)):
    input_x=sequence[:i]
    output_y=sequence[i]
    X.append(input_x)
    Y.append(output_y)
Y
```

```python
X
```

```python
len(X)
```

```python
maxlen=max([len(s) for s in sequences])
print(maxlen)
X_padded=pad_sequences(sequences,maxlen=maxlen,padding="post")
print(X_padded.shape)
X_padded
```

```python
Y=np.array(Y)
```

```python
embed_size=100
units=256
```

# Training

RNN

```python
epochs=50
batch_size=32
```

```python
rnn_model=Sequential([
    Embedding(input_dim=len(tokenizer.word_index)+1,output_dim=embed_size,mask_zero=True),   # +1 for padded 0
    SimpleRNN(units=units),
    Dense(units=len(tokenizer.word_index)+1,activation="linear")
])
rnn_model.compile(loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),optimizer="adam",metrics=["accuracy"])
history_rnn=rnn_model.fit(X_padded,Y,epochs=epochs,batch_size=batch_size,validation_split=0.2)
```

LSTM

```python
lstm_model=Sequential([
    Embedding(input_dim=len(tokenizer.word_index)+1,output_dim=embed_size,mask_zero=True),   # +1 for padded 0
    LSTM(units=units),
    Dense(units=len(tokenizer.word_index)+1,activation="linear")
])
lstm_model.compile(loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),optimizer="adam",metrics=["accuracy"])
history_lstm=lstm_model.fit(X_padded,Y,epochs=epochs,batch_size=batch_size,validation_split=0.2)
```

GRU

```python
from tensorflow.keras.layers import GRU

gru_model = Sequential([
    Embedding(
        input_dim=len(tokenizer.word_index)+1,
        output_dim=embed_size,
        mask_zero=True
    ),
    GRU(units=units),
    Dense(len(tokenizer.word_index)+1, activation="linear")
])

gru_model.compile(
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    optimizer="adam",
    metrics=["accuracy"]
)

history_gru = gru_model.fit(
    X_padded,
    Y,
    epochs=epochs,
    batch_size=batch_size,
    validation_split=0.2
)
```
